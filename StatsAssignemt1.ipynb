```python
print("Hello, Woodmill!")

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("All libraries imported successfully!")
```

```python
d=pd.read_excel("WoodMill.xlsx")
print(d.head(3))
```

```python
print(d.head(),"\n")
print(d.describe())

d.columns
```

```python
##in the above block of code I removed the first row because it doesnt have any numerical value in it 
d=d.iloc[1:].copy()
```

```python
d=d.iloc[1:].copy()
d["Scanner"]=pd.to_numeric(d["Scanner"])
d["Manual"]=pd.to_numeric(d["Manual"])
print(d.head())
print(d.describe())
```

```python
bins=[5,15,25,35,45,55]
scanner_freq=pd.cut(d["Scanner"],bins=bins,right=False).value_counts().sort_index()
manual_freq=pd.cut(d["Manual"],bins=bins,right=False).value_counts().sort_index()

frequency_table=pd.DataFrame({"Scanner_Freq":scanner_freq,"Manual_Freq":manual_freq})

print(frequency_table)
```

```python
print(d.describe())

##Range = max-min
scanner_range=d["Scanner"].max()-d["Scanner"].min()
manual_range=d["Manual"].max()-d["Manual"].min()

##Coefficient of variance spread(standard Deviation) relative to the mean value 
scanner_cv=( d["Scanner"].std()/d["Scanner"].mean())*100
manual_cv=( d["Manual"].std()/d["Manual"].mean())*100

print("Scanner Range",scanner_range,"\n")
print("Manual Range",manual_range,"\n")

print("Scanner Coefficien of Variance",scanner_range,"\n")
print("Manual  Coefficien of Variance",manual_range,"\n")

#now lets make a table 

summary=pd.DataFrame({
    "Scanner":d["Scanner"].describe(),
    "Manual":d["Manual"].describe()
})

#the .loc adds a row called range and cv and the placeholders are the values i provide 
summary.loc["Range"]=[scanner_range,manual_range]
summary.loc["Coefficient Of Variance"]=[scanner_cv,manual_cv]

print("\n",summary.round(2))
```
