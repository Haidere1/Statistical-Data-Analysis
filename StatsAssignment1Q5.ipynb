Question 5: First I will check the total number of observations in the DataSet (column: Total) 

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
Orlandao_AirportDS = pd.read_excel("OrlandoAirport_fixed.xlsx",header=2)

##the excel file is a bit messed up
#I Used header = 2 so total gets recongnized as the column name and i can use it later on

Orlandao_AirportDS.head()
```

```python
Orlandao_AirportDS["Total"]
```

```python
bin = np.linspace(Orlandao_AirportDS["Total"].min(),Orlandao_AirportDS["Total"].max(),5)
freq= pd.cut(Orlandao_AirportDS["Total"],bins=bin,include_lowest=True).value_counts().sort_index()

freq.plot(kind="bar",figsize=(10,12))
plt.title("Grouped Frequency Distribution")
plt.ylabel("Frequency")
plt.xlabel("Passenger Total Interval")
plt.xticks(rotation=45, ha="right")
# I used the following line to get an accurate display of the frequencies on the Yaxis
plt.yticks(range(0, int(freq.max()) + 1))
plt.tight_layout()
plt.show()
```
